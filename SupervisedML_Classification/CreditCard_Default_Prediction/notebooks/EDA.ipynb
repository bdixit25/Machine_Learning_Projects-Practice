### Dataset Information & Background

> This research investigates customer default payments in Taiwan, comparing the predictive accuracy of the probability of default across six different data mining methods. 

* **Objective:** From a risk management perspective, estimating the actual probability of default provides more value than a binary classification (credible vs. not credible).
* **Methodology:** Since the true probability of default is inherently unknown, the study introduces the **Sorting Smoothing Method** to estimate it. 
* **Variables:** 
  * **Response Variable ($Y$):** The estimated real probability of default.
  * **Independent Variable ($X$):** The predictive probability of default.

---

### Key Findings
* **Simple Linear Regression Result:** $Y = A + BX$
* **Best Performing Model:** **Artificial Neural Network (ANN)** produced the highest coefficient of determination among the six evaluated data mining techniques. 
* **Model Performance:** ANN is the only technique where the regression intercept ($A$) is close to zero and the regression coefficient ($B$) is close to one, successfully and accurately estimating the real probability of default.
